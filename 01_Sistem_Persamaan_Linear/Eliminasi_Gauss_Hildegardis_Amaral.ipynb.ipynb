## 1. Penjelasan Metode Eliminasi Gauss

Metode Eliminasi Gauss merupakan salah satu metode numerik yang digunakan untuk menyelesaikan Sistem Persamaan Linear (SPL).

Metode ini bekerja dengan mengubah sistem persamaan menjadi bentuk matriks segitiga atas melalui operasi baris elementer. Setelah itu, nilai setiap variabel dicari menggunakan proses substitusi mundur.

Tahapan penyelesaian:

1. Menyusun sistem persamaan ke dalam bentuk matriks augmented.
2. Melakukan eliminasi pada elemen di bawah diagonal utama.
3. Mengubah matriks menjadi bentuk segitiga atas.
4. Melakukan substitusi mundur untuk memperoleh nilai variabel x, y, dan z.
5. Menampilkan hasil akhir perhitungan.

```python

# Cell 2: Memasukkan matriks koefisien dan konstanta

A = np.array([
    [4,  2, -2],
    [2,  6,  4],
    [3, -1,  2]
], dtype=float)

B = np.array([5, 35, 10], dtype=float)

# Menampilkan matriks augmented
matriks_augmented = np.column_stack((A, B))

print("Matriks Augmented:")
print(matriks_augmented)
```

```python

# Cell 3: Proses Eliminasi Gauss

# Menggabungkan matriks koefisien dan konstanta
M = np.column_stack((A.copy(), B.copy()))

# Menentukan jumlah variabel
n = len(B)

# Melakukan proses eliminasi
for k in range(n - 1):

    # Memilih baris dengan nilai pivot terbesar
    baris_pivot = k + np.argmax(np.abs(M[k:, k]))

    # Menukar baris jika diperlukan
    if baris_pivot != k:
        M[[k, baris_pivot]] = M[[baris_pivot, k]]

    # Menghilangkan elemen di bawah pivot
    for i in range(k + 1, n):
        faktor = M[i, k] / M[k, k]

        for j in range(k, n + 1):
            M[i, j] = M[i, j] - faktor * M[k, j]

    print("Matriks setelah eliminasi kolom", k + 1)
    print(M)
    print()
```

```python

# Cell 4: Proses substitusi mundur

# Menyiapkan tempat untuk menyimpan hasil variabel
X = np.zeros(n)

# Menghitung nilai variabel dari bawah ke atas
for i in range(n - 1, -1, -1):

    jumlah = 0

    for j in range(i + 1, n):
        jumlah += M[i, j] * X[j]

    X[i] = (M[i, n] - jumlah) / M[i, i]

# Menampilkan hasil setiap variabel
print("Hasil Penyelesaian SPL:")
print("Nilai x =", X[0])
print("Nilai y =", X[1])
print("Nilai z =", X[2])
```

```python

# Cell 5: Memeriksa hasil penyelesaian

# Menghitung kembali nilai ruas kiri persamaan
hasil_pemeriksaan = np.dot(A, X)

# Membandingkan hasil dengan konstanta
print("Pemeriksaan hasil perhitungan:")
print("Hasil ruas kiri:", hasil_pemeriksaan)
print("Nilai konstanta:", B)

# Memeriksa apakah hasil sudah sesuai
if np.allclose(hasil_pemeriksaan, B):
    print("Kesimpulan: Solusi benar dan memenuhi semua persamaan.")
else:
    print("Kesimpulan: Solusi belum memenuhi semua persamaan.")
```
